```python
section_number = rows[0]
section_name = rows[1]
chunk = []

for i in range(len(index_articles)):
    
    
    if i == (len(index_articles) - 1): 
        article_title = rows[index_articles[i]:]
        article_no = article_title[0].split()[-1]        
        body = article_body(article_title)

        for govde in body:
            chunk_dict = {}
            govde_value = govde.split(".")[0]
            chunk_dict["ID"] = f"{article_no}({govde_value})"
            chunk_dict["Article_No"] = article_no
            chunk_dict["Article_title"] = article_title[1]
            chunk_dict["Body_No"] = govde_value
            chunk_dict["Body"] = govde
            chunk_dict["References"] = f"Regulation (EU) 2024/1689, Chapter III, Section 2, Article {article_no}({govde_value})"
            chunk.append(chunk_dict)
        
    else:
        article_title = rows[index_articles[i]:index_articles[i+1]]
        article_no = article_title[0].split()[-1]   
        body = article_body(article_title)

        for govde in body: 
            chunk_dict = {}
            govde_value = govde.split(".")[0]
            chunk_dict["ID"] = f"{article_no}({govde_value})"
            chunk_dict["Article_No"] = article_no
            chunk_dict["Article_title"] = article_title[1]
            chunk_dict["Body_No"] = govde_value
            chunk_dict["Body"] = govde
            chunk_dict["References"] = f"Regulation (EU) 2024/1689, Chapter III, Section 2, Article {article_no}({govde_value})"
            chunk.append(chunk_dict)
            
print(chunk)
```

```python
def article_body(article):
    govde_index = []
    for i, x in enumerate(article):
        if x.strip() and x.strip()[0].isdigit():
            govde_index.append(i)
    govde = []
    for i in range(len(govde_index)):
        if i == len(govde_index) - 1:
            sonuc = " ".join(article[govde_index[i]:])
            govde.append(sonuc)

        else:
            sonuc = " ".join(article[govde_index[i]: govde_index[i+1]])
            govde.append(sonuc)

    return govde 
```

```python
def search_query(s, k=5):

    embeddings_s = model.encode(s, normalize_embeddings=True)

    result = np.dot(embeddings, embeddings_s)
    scores = {}
    for c, i in zip(chunk, result):
        scores[c["ID"]] = float(i)
    
    top_k = sorted(scores.items(), key=lambda item: item[1], reverse=True)[:k]

    return top_k
```

```python
def build_lookup(x):

    lookup_dict = {c["ID"]: c["Body"] for c in x}
    
    return lookup_dict
```

```python
def answer(q, k):
    
    top_k = search_query(q, k)
    
    context = build_context(top_k, lookup_dictionary)

    message = client.messages.create(max_tokens=3500, model="claude-opus-5", messages=[{"role":"user", "content":context+" "+q}], system=system_prompt)
    
    return message, top_k
```

```python
import time 
def run(questions, k):

    run_q = []
    
    for i in questions: 
        results = {}
        
        results["Question_ID"] = i
        results["Question_Text"] = questions[i]

        start_time = time.time()
        message, top_k = answer(questions[i], k)
        end_time = time.time()

        results["Chunk_ID"] = top_k
        
        for j in message.content:
            if j.type=="text":
                results["Answer_Text"] = j.text
        
        results["Input_Tokens"] = message.usage.input_tokens
        results["Output_Tokens"] = message.usage.output_tokens
        results["Stop_Reason"] = message.stop_reason
        results["Time"] = end_time - start_time
        run_q.append(results)
    return run_q
```

```python
def scoring(s):

    retrieval = {}
    for i in s: 
    
        if i["Question_ID"] in gold:
            retrieval[i["Question_ID"]] = {x[0] for x in i["Chunk_ID"]}
            
    scores = {}
    for k in gold:
        intersection = gold[k] & retrieval[k]
        missed  = gold[k] - retrieval[k]
        recall  = len(intersection) / len(gold[k])
        scores[k] = {"recall": recall, "missed": missed}
    return scores
```

```python
def extract_citations(text):
    with_para = set(re.findall(r"\d+\(\d+\)", text))
    bare_raw  = set(re.findall(r"Articles?\s+(\d+)", text))

    para_articles = {x.split("(")[0] for x in with_para}         

    bare = bare_raw - para_articles
    return with_para, bare
```

```python
sys_wp, sys_bare = extract_citations(system_prompt)
citations_dict = {}
for i in runs10:
    wp, bare = extract_citations(i["Answer_Text"])
    chunk_ids = {x[0] for x in i["Chunk_ID"]}
    suspect = wp - chunk_ids
    
    bodies = " ".join(lookup_dictionary[c] for c in chunk_ids)
    body_wp, body_bare = extract_citations(bodies)
    fake      = suspect - (body_wp   | sys_wp)     
    bare_fake = bare    - (body_bare | sys_bare)    
    citations_dict[i["Question_ID"]] = {"with_para": wp, "bare": bare, "suspect": suspect, "fake": fake, "bare_fake":bare_fake}


for q, v in citations_dict.items():
    print(q, "| suspect:", sorted(v["suspect"]), "| fake:", sorted(v["fake"]), "| bare fake", sorted(v["bare_fake"]))
```

```python
def correctness_table(d):
    for q in sorted(d):
        payda = d[q]["payda"]
        satir = [f"{q}  payda={payda}"]
        for kk in ("k5", "k10"):
            if kk in d[q]:
                v = d[q][kk]
                satir.append(f"{kk}: {v['pay']}/{payda} = {v['pay']/payda:.2f}  not_ok={v['not_ok']}")
            else:
                satir.append(f"{kk}: —")
        print(" | ".join(satir))

correctness_table(correctness)
```

```python
def score_table(runs, correctness, behaviour, kk, retrieval_scores, citations):
    print(f"\n{'='*72}\n  k = {kk[1:]}\n{'='*72}")
    print(f"{'QUESTION':<6}{'Retrieval':>12}{'Groundedness':>16}{'Correctness':>16}   Missed Chunk")
    print("-"*72)

    for q in ["A1","A2","A3","A4","B1","B2","B3"]:
        r = retrieval_scores[q]
        c = correctness.get(q, {})
        v = c.get(kk)

        ret = f"{r['recall']:.2f}"
        gnd = "CLEAN" if not (citations[q]["fake"] or citations[q]["bare_fake"]) else "FLAG"
        cor = f"{v['pay']}/{c['payda']}" if v else "—"
        kacan = ", ".join(sorted(r["missed"])) or "-"

        print(f"{q:<6}{ret:>12}{gnd:>16}{cor:>16}   {kacan}")

    print("-"*72)
    for q in ["C1","C2","C3","C4"]:
        gnd = "CLEAN" if not (citations[q]["fake"] or citations[q]["bare_fake"]) else "FLAG"
        dav = "REJECTED ✓" if behaviour[q][kk] else "ANSWERED ✗"
        print(f"{q:<6}{'(NO GOLD)':>12}{gnd:>16}{dav:>16}")

score_table(runs5,  correctness, behaviour, "k5",  runs_5_scoring,  citations_dict_5)
score_table(runs10, correctness, behaviour, "k10", runs_10_scoring, citations_dict)
```
